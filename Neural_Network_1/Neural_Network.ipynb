```python
X_train = X_train / 255
X_test = X_test / 255


X_train_flattened  = X_train.reshape(len(X_train), 28*28)
X_train_flattened
```

```python
X_test_flattend = X_test.reshape(len(X_test), 28*28)
X_test_flattend
```

```python
X_test.shape   # when you multiply the 28  with 28 then you will get the answer of 784
```

```python
# now we will create the neural network for that
model = keras.Sequential([
    keras.layers.Dense(10, input_shape =(784, ), activation  = 'sigmoid')
])


model.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
              )

model.fit(X_train_flattened, y_train, epochs = 5)
```

now we will scale the value and then we will see the accuracy

```python
# now we will create the neural network ... these are after scaling result
model = keras.Sequential([
    keras.layers.Dense(10, input_shape =(784, ), activation  = 'sigmoid')
])


model.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
              )

model.fit(X_train_flattened, y_train, epochs = 5)
```

```python
# Now we will evaluate the model
model.evaluate(X_test_flattend, y_test)
```

```python
plt.matshow(X_test[1])
```

```python
# now we test
y_pred = model.predict(X_test_flattend)
y_pred[1]
```

```python
print('By here you can evaluate the model performance once you check out and compare it with above',np.argmax(y_pred[1]))
```

```python
y_predicted_label = [np.argmax(i) for i in y_pred]
y_predicted_label[:5]
```

```python
cm = tf.math.confusion_matrix(labels = y_test, predictions = y_predicted_label)
```

```python
cm
```

```python
type(cm)
```

```python
sns.heatmap(
    cm,
    annot = True,
    fmt = 'd'
)
plt.xlabel('Truth ')
plt.ylabel('Predicted')
```

```python
# now we will add the hidden layer to the model  once we will add the hidden layer it will improve the performance

model = keras.Sequential([
    keras.layers.Dense(100, input_shape =(784, ), activation  = 'relu'),
    keras.layers.Dense(10, activation  = 'sigmoid')
])


model.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
              )

model.fit(X_train_flattened, y_train, epochs = 5)
```

```python
# now we will add the hidden layer to the model  once we will add the hidden layer it will improve the performance

model = keras.Sequential([
    keras.layers.Flatten(input_shape = (28, 28)),
    keras.layers.Dense(100, activation  = 'relu'),
    keras.layers.Dense(10, activation  = 'sigmoid')
])


model.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
              )

model.fit(X_train,  y_train, epochs = 5)
```
